- CSV 파일 DataFrame 으로 읽기

```python
import pandas as pd
import numpy as np

path = '/content/drive/MyDrive/Colab Notebooks/diabetes.csv'  # 읽으려는 CSV 파일 경로 복사 및 붙여놓기

df = pd.read_csv(path)   # CSV 파일을 DataFrame으로 읽기

df
```

- 정답/입력데이터 설정

```python
# 'Outcome' 컬럼 제거
df2=df.drop('Outcome',axis=1)
df.head()

# y값(정답) 지정
y=df2['BMI']
y.value_counts()
y

# X값(입력데이터) 지정 : 세로축을 기준으로 BMI 한 줄을 제외한 나머지 데이터
X=df2.drop('BMI',axis=1)
X.head()
```

- 학습데이터/테스트데이터 분할

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.25,random_state=0)
```

- 모델 import

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

- Linear Regression

```python
model = LinearRegression()
model.fit(X_train, y_train)
ly_preds = model.predict(X_test)

print('평균제곱근오차', mean_squared_error(ly_preds, y_test))

def mse_np(actual, predicted):
    return np.mean((np.array(actual) - np.array(predicted)) ** 2)

print('평균제곱근오차', mse_np(ly_preds, y_test))

def mse(actual, predicted):
    sum_square_error = sum((a - p) ** 2for a, p in zip(actual, predicted))
    mean_square_error = sum_square_error / len(actual)
    return mean_square_error

print('평균제곱근오차', mse(ly_preds, y_test))
```

- Decision Tree Regression

```python
model = DecisionTreeRegressor()
model.fit(X_train, y_train)

from sklearn.metrics import mean_squared_error

dy_preds = model.predict(X_test)
print('평균제곱근오차', mean_squared_error(dy_preds, y_test))
```

- Random Forest Regression

```python
model = RandomForestRegressor()
model.fit(X_train, y_train)

from sklearn.metrics import mean_squared_error

ry_preds = model.predict(X_test)
print('평균제곱근오차', mean_squared_error(ry_preds, y_test))
```
